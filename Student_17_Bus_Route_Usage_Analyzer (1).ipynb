## 2. Load your dataset

```python
df = pd.read_csv("Student_17_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# TODO: Display the shape and column names
# TODO: Inspect data types and missing values
# TODO: Count duplicate rows
print('First five rows:')
display(df.head())
print('\nShape (rows, columns):', df.shape)
print('Column names:', list(df.columns))
print('\nData types:')
print(df.dtypes)
print('\nMissing values in each column:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
passengers = df['Daily_Passengers'].to_numpy()

average_passengers = np.mean(passengers)
maximum_passengers = np.max(passengers)
minimum_passengers = np.min(passengers)

print('Average passengers :', average_passengers)
print('Maximum passengers :', maximum_passengers)
print('Minimum passengers :', minimum_passengers)

# Second NumPy calculation: average route-stop distance
distances = df['Distance_KM'].to_numpy()
print('Average distance (km):', np.mean(distances))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
df['Passengers_Per_KM'] = np.where(df['Distance_KM'] != 0, df['Daily_Passengers'] / df['Distance_KM'], np.nan)


df['Passenger_Category'] = np.where(df['Daily_Passengers'] >= df['Daily_Passengers'].mean(), 'High', 'Regular')

print('Records with passenger count above the dataset average:')
display(df[df['Daily_Passengers'] > df['Daily_Passengers'].mean()])

print('Top 5 records by daily passengers:')
display(df.sort_values(by='Daily_Passengers', ascending=False).head(5))

print('Top 5 records by passengers per kilometre:')
display(df.sort_values(by='Passengers_Per_KM', ascending=False).head(5))
```

## 6. Your individual project challenge
**Challenge:** Calculate passenger totals by route; find busiest stop; calculate passengers per kilometre.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Passenger totals for each route
route_totals = df.groupby('Route', as_index=False)['Daily_Passengers'].sum()
route_totals = route_totals.sort_values('Daily_Passengers', ascending=False)
print('Total passengers by route:')
display(route_totals)

# Busiest stop 
stop_totals = df.groupby('Stop', as_index=False)['Daily_Passengers'].sum()
busiest_stop = stop_totals.sort_values('Daily_Passengers', ascending=False).iloc[0]
print('Busiest stop:', busiest_stop['Stop'])
print('Passengers recorded at busiest stop:', busiest_stop['Daily_Passengers'])

# Passengers per kilometre 
print('Record with the highest passengers per kilometre:')
display(df.sort_values('Passengers_Per_KM', ascending=False).head(1))
route_efficiency = df.groupby('Route', as_index=False)['Passengers_Per_KM'].mean().sort_values('Passengers_Per_KM', ascending=False)
print('Average passengers per kilometre by route:')
display(route_efficiency)
```
